```python
import math


mass = 0.15
radius = 0.60
frequency = 2.0


T = 1 / frequency
print(f"(a) 주기: {T:.2f} s")


v = 2 * math.pi * radius / T
print(f"(b) 속력: {v:.2f} m/s")


a_c = v**2 / radius
print(f"(c) 구심가속도: {a_c:.2f} m/s²")
```

```python
import math


theta_deg = 45
theta_rad = math.radians(theta_deg)
g = 9.8

tan_theta = math.tan(theta_rad)

v_squared_over_r = g * tan_theta
print(f"(1) v² / r = {v_squared_over_r:.2f} (m²/s²)/m")

a_c = v_squared_over_r
print(f"(2) 구심가속도: {a_c:.2f} m/s²")
```

```python

v_initial = 0
v_final = 32
time = 10.5
radius = 500
v_current = 35

a_t = (v_final - v_initial) / time
print(f"(a) 접선 가속도: {a_t:.2f} m/s²")

a_r = v_current**2 / radius
print(f"(b) 지름 가속도: {a_r:.2f} m/s²")
```
